## 7. Memory measurement cross-check with memory_profiler
`memory_profiler` samples the whole process, so it is coarse for tiny functions. This is why the project uses `tracemalloc` for the main results.

```python
from memory_profiler import memory_usage

data = sa.generate_sorted_dataset(50000)
usage = memory_usage((sa.linear_search, (data, sa.worst_case_key(50000))), interval=0.001)
print("Linear Search on 50000 items - process memory while running (MiB): min %.2f, max %.2f" % (min(usage), max(usage)))
```
